# Bitácoras Onmásticas del SEMEFO-DF de 1968-1982
# Generación de profiles masiva

```python
import os
from glob import glob

import pandas as pd
from ydata_profiling import ProfileReport

target_dir = "BO_20230222"
profile_minimal = True
profile_force_refresh = False

#Obtener archivos 
in_files = glob( "data_raw/{}/*.xlsx".format(target_dir) )
in_files

# check if file exists
out_dir = "profiles/{}/".format(target_dir)
check_dir = os.path.isdir(out_dir)
if not check_dir:  os.makedirs(out_dir)

# read every file
for in_file in in_files:    
    print("Reading: "+in_file)
    try: df = pd.read_excel(in_file, dtype='str')
    except: print("Failed to read file: "+in_file)
    else:
        out_file = in_file.replace("data_raw/","profiles/").replace(".xlsx",".html")
        if not profile_force_refresh and os.path.exists(out_file):
            print("File exists: "+out_file)
        else:
            print("Creating: "+out_file)
            prof = ProfileReport(df, title=in_file, minimal=profile_minimal) 
            prof.to_file(output_file=out_file)
```

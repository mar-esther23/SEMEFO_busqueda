```python
import pandas as pd
from numpy import nan
import datetime

import matplotlib.pyplot as plt
import seaborn as sns

def summary_top_series(s, n_values=5, sep=', ', **kwargs):
    s = s.value_counts(**kwargs).head(n_values)
    s = s.to_dict()
    s = sep.join([f"{k} ({v})" for k,v in s.items()])
    return s
```

## 3. Eventos de interés

```python
date_min = datetime.datetime(1968, 10, 2)
date_max = date_min + datetime.timedelta(days=7)

data = df.loc[df['Fecha_estandar'].between(date_min, date_max, inclusive='both'), key_col].sort_values(by=['Fecha_estandar','Procedencia_acta'])
data
```

```python
date_min = datetime.datetime(1975, 4, 25)
date_max = date_min + datetime.timedelta(days=7)

data = df.loc[(df['Fecha_estandar'].between(date_min, date_max, inclusive='both')) & (df['Diagnostico_estandar'].str.contains('HPAF')==True), key_col] \
         .sort_values(by=['Fecha_estandar','Procedencia_acta'])
data
```

## 4. Grupos de interés
### Por acta

```python
data = df.loc[df['Diagnostico_estandar'].str.contains('HPAF')==True, key_col]
display(data.shape)
index = data['Procedencia_acta'].value_counts()>2
index = index[index].index
data[data['Procedencia_acta'].isin(index)].sort_values(by='Procedencia_acta')
```

### Por día de ingreso

```python
data = df.loc[df['Diagnostico_estandar'].str.contains('HPAF')==True, key_col]
data.groupby('Fecha_estandar').filter(lambda g: len(g)>3).sort_values(by=['Fecha_estandar','Procedencia_acta'])
```

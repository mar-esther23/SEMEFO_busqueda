### Fecha_diasemana

```python
df['Fecha_estandar'] = pd.to_datetime(df['Fecha_transcrito'], errors='coerce')
df['Fecha_estandar'].value_counts(dropna=False)
```

```python
df['Fecha_año'] = df['Fecha_estandar'].dt.year.fillna('') \
                  .apply(lambda s: str(int(s)) if s!='' else nan)
display( df['Fecha_año'].value_counts(dropna=False).to_dict() )

df['Fecha_mes'] = translate_datetime(df['Fecha_estandar'].dt.month_name())
display( df['Fecha_mes'].value_counts(dropna=False).to_dict() )

df['Fecha_añomes'] = df[['Fecha_año','Fecha_mes']] \
                       .apply(join_strings_with_nans, sep=' ', axis=1)
df = df.drop(['Fecha_año','Fecha_mes'], axis=1)
df['Fecha_añomes'].value_counts(dropna=False)
```

```python
replace = {'1970-04-31': '1970 Abril', '1970-02-29': '1970 Febrero', '1971-02-29': '1971 Febrero', '1972-02-30': '1972 Febrero', '1974-02-29': '1974 Febrero', '1976-04-31': '1976 Abril', '1976-11-31': '1976 Noviembre', '1976-06-31': '1976 Junio', '1976-02-31': '1976 Febrero', '1978-02-30': '1978 Febrero', '1978-02': '1978 Febrero', '1977-12': '1977 Diciembre', '1978-04': '1978 Abril', '1978-05': '1978 Mayo', '1979-11-31': '1979 Noviembre', '1979-02-29': '1979 Febrero', '1979-02-30': '1979 Febrero', '1979-02-31': '1979 Febrero', '1979-08': '1979 Agosto', '1979-09': '1979 Septiembre', '1979-01': '1979 Enero', '1980-06-31': '1980 Junio', '1981-02-31': '1981 Febrero', '31/04/1981': '1981 Abril', }
df['Fecha_añomes'] = df['Fecha_añomes'].fillna(df['Fecha_transcrito']).replace(replace)
df.loc[(df['Fecha_estandar'].isna()) & (df['Fecha_añomes']!='S-D'),['ID','Fecha_transcrito','Fecha_añomes']]
```

```python
año = [str(i) for i in range(1968, 1983)]
mes = ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo', 'Junio', 'Julio', 
       'Agosto', 'Septiembre', 'Octubre', 'Noviembre', 'Diciembre', ]
cat_type = ['S-D'] + [' '.join(p) for p in product(año, mes)]
cat_type = pd.CategoricalDtype(categories=cat_type, ordered=True)

df['Fecha_añomes'] = df['Fecha_añomes'].astype(cat_type)
df['Fecha_añomes'].value_counts(dropna=False)
```

```python
df['Fecha_diasemana'] = translate_datetime(df['Fecha_estandar'].dt.day_name())
dia = ['Lunes', 'Martes', 'Miércoles', 'Jueves', 'Viernes', 'Sábado', 'Domingo']
cat_type = pd.CategoricalDtype(categories=dia, ordered=True)
df['Fecha_diasemana'] = df['Fecha_diasemana'].astype(cat_type)
df['Fecha_diasemana'].value_counts(dropna=False)
```

## Edad_transcrito
### Edad_estandar
### Edad_rango

```python
from math import floor
from helper_pandas import to_dtype_with_errors

def replace_edades_menor_al_año(edad):
    if type(edad)==str:
        edad_ = simplify_string(edad)
        edad_ = edad_.replace(' 1/2','.5')
        edad_ = edad_.replace('horas','hora').replace('dias','dia').replace('semanas','semana').replace('meses','mes').replace('anos','ano')
        dic_time = {'ano':1, 'mes':12, 'semana':52, 'dia':365, 'hora':8766}
        for nombre, n in dic_time.items():
            if nombre in edad_:
                edad_ = simplify_string( edad_.replace(nombre,'') )
                try: edad_ = float(edad_)
                except ValueError: return edad
                return floor( edad_ / n )
        dic_time = {'a':1, 'm':12, 'd':365}
        for nombre, n in dic_time.items():
            if nombre in edad_:
                edad_ = simplify_string( edad_.replace(nombre,'') )
                try: edad_ = float(edad_)
                except ValueError: return edad
                return floor( edad_ / n )
        try: edad_ = float(edad_)
        except(ValueError): return edad
        return floor(edad_)
    return edad

dic_replace = {'[1-10 días]':0, '10 SEMNS':0, '15-30 días':0, '2 mese':0, '2 o 3 meses':0, '30 - 90 días':0, '5 meses 15 días':0, 
               '6 meses UI':0, '7 - 12 meses':0, '72 hrs':0, '-12 horas':1, '1 año 3 meses':1, '1 año 4 meses':1, '1 año 7 meses':1, 
               '1 Y 8 meses':1, '15 mese':1, '18 mese':1, '3 años y medio ':3, '7 añoss':7, '[8]':8, '-72 años':72, ' S-D':'S-D', 
               '-':'S-D', '?':'S-D', 'F':'S-D', 'M':'S-D', 'S-d':'S-D', '-72 años':72, '-1':1, '2/6':'S-D', '$45':45, '3-5':4, 
               ' +3-5':4,'7-9':8,' +6-8':7,' +7-9':8,'4-6':5,'5-7':6,'6-8':7,'4 y 5':'S-D',' +4-6':5,'8 años7':8,'más 72':'S-D',
               '1 - 3':2,'1-2':1,'9-9':9,'3-4':3,'7-8':7,'2-4':3,'8 Ins.':8,'41 n':41,'R.N':0,'18-25':21, '7 - 9':8,
               '25-30':27,' +3-4':3,' +5-6':5,' +2-3':2,' +4-7':5,' +7-8':7,' +5-7':6,' +6-7':6,'25-40':'S-D','20 o 24':22,'25/30':22,
               #'8 172':'S-D',
              }

df['Edad_estandar'] = df['Edad_transcrito']
df['Edad_estandar'] = df['Edad_estandar'].apply(replace_edades_menor_al_año) \
                       .replace(',','.', regex=True).replace(dic_replace)
df = to_dtype_with_errors(df, 'Edad_estandar')
df.loc[df['Tipo_restos'].isin(['Feto','Recién nacido']), 'Edad_estandar_clean'] = 0
df['Edad_estandar'] = df['Edad_estandar_clean']
df = df.drop(['Edad_estandar_clean', 'Edad_estandar_error'], axis=1)
df['Edad_estandar'].value_counts()
```
